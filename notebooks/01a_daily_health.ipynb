## 01a — Daily Health

This notebook runs and validates the SQL analysis for Part 1(a) of the assignment.

Metrics are calculated by day and platform:
- Daily Active Players
- Gross Revenue
- ARPDAU
- Share of active players who made a purchase that day

```python
import sys
sys.path.append("..")

import pandas as pd
from src.db import get_connection

conn = get_connection()

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)

print("Database connected successfully.")
```

```python
sql_path = "../sql/1a_daily_health.sql"

with open(sql_path, "r", encoding="utf-8") as f:
    query = f.read()

df_01a = pd.read_sql_query(query, conn)

print("Shape:", df_01a.shape)
df_01a.head(10)
```

```python
print("Rows:", len(df_01a))
print("Columns:", df_01a.columns.tolist())

print("\nMissing values:")
print(df_01a.isna().sum())

print("\nDuplicate day-platform rows:")
print(df_01a.duplicated(subset=["date", "platform"]).sum()
      if {"date", "platform"}.issubset(df_01a.columns)
      else "Check actual date column name")
```

```python
print("Daily Health Summary")
display(df_01a.describe(include="all"))

# ARPDAU should equal gross revenue / DAU
if {"gross_revenue", "dau", "arpdau"}.issubset(df_01a.columns):
    expected_arpdau = df_01a["gross_revenue"] / df_01a["dau"]

    max_diff = (df_01a["arpdau"] - expected_arpdau).abs().max()

    print("\nMaximum ARPDAU calculation difference:", max_diff)
```

```python
output_path = "../outputs/tables/01a_daily_health.csv"

df_01a.to_csv(output_path, index=False)

print(f"Saved successfully: {output_path}")
```

```python
saved = pd.read_csv("../outputs/tables/01a_daily_health.csv")

print("Saved shape:", saved.shape)
print("Current shape:", df_01a.shape)

print(
    "Output file verified:",
    len(saved) == len(df_01a)
    and saved.columns.tolist() == df_01a.columns.tolist()
)

saved.head()
```
